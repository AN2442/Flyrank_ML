## 1. My lane as an ML task (type)

**Lane:** Refresh / Content Opportunity Scoring

**ML task type:** Scoring

I frame this lane as a scoring problem because the goal is to assign each page a priority score based on observable page and performance signals. The scores can then be used to create a ranked list of pages for content review.

The output is not an automatic refresh decision. Instead, it is a decision-support signal that helps the content/SEO team decide which pages to review first.

 Given the available page-level signals, estimate a priority score for each page that can be used to rank pages for potential content review or refresh.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


There is no observed label in the starter dataset that says whether a page actually needed a content refresh. Therefore, I would use an observed performance outcome as a proxy rather than creating a refresh label from my own rule.

**Target/proxy:** `trend_pct`

`trend_pct` is an observed page-level performance-change signal. I would use it as a proxy for deterioration or improvement in page performance, where more negative values indicate stronger observed decline.

The scoring output would use the learned relationship between page characteristics and this observed outcome to prioritize pages for review.

This is only a proxy for refresh opportunity. A negative trend does not prove that a page needs a refresh, and it does not establish that content age or any other factor caused the decline. The final decision remains with the content/SEO team.

I would also keep `trend_pct` and `trend_direction` out of the model features because they are outcome-derived fields rather than independent predictors.

## 3. Success metric

*One metric you can defend. What number means 'good'?*


**Primary metric: Precision@50**

I would use Precision@50 because the practical decision is to identify which pages should be reviewed first, rather than treating every page equally.

Precision@50 measures the proportion of the 50 highest-priority pages that match the observed target/proxy.

For example, if 40 of the top 50 ranked pages match the target condition, Precision@50 would be 0.80.

This metric fits the real content-review workflow because the team has limited capacity and is more likely to act on a small prioritized queue than review every page.

A higher Precision@50 would mean that more of the pages placed at the top of the review queue correspond to the outcome being used as the proxy. The metric therefore evaluates whether the scoring approach is useful for prioritizing pages for human review.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*



A fixed rule could prioritize pages using manually chosen thresholds, for example by selecting pages that are both relatively old and showing a negative performance trend. This approach is transparent and useful as a baseline, but the thresholds and relative importance of the signals would be chosen manually.

ML is worth testing because page characteristics and performance signals may interact in ways that are difficult to capture with a small number of fixed thresholds. A learned scoring approach can use multiple observed features to estimate a priority score and rank pages for review.

The goal is not to assume that ML is automatically better. The ML approach should be compared against a simple fixed-rule baseline using the same evaluation metric, Precision@50.

If the learned approach produces a more useful top-50 review queue than the baseline, that would provide evidence that the additional complexity is justified. If it does not, the simpler rule may be preferable.

The final output remains decision support: the model ranks pages for human review rather than automatically deciding that a page needs to be refreshed.
