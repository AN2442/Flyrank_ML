## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

url = "https://raw.githubusercontent.com/AN2442/Flyrank_ML/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Number of pages:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print(df["trend_direction"].value_counts())
```

```python
age_by_trend = df.groupby("trend_direction")["content_age_days"].median()

print(age_by_trend)
```

```python
print("Pages by trend direction:")
print(df["trend_direction"].value_counts())

print("\nPercentage of pages that are declining:")
print(round((df["trend_direction"].eq("down").mean()) * 100, 2), "%")
```

Content age differs across trend directions in this dataset, which makes it worth investigating whether content age and other page signals can help prioritize pages for review.

### What I found

The starter dataset contains many pages that can be compared using signals such as content age and trend direction. The median content age differs across trend directions, which gives me a reason to investigate whether page characteristics can help identify pages that deserve review.

These numbers do not prove that content age causes a page's trend. They only show that there is a pattern worth investigating.
